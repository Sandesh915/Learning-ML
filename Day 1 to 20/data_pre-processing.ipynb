For Numerical Column Encoding

```python
from sklearn.impute import SimpleImputer

numeric_columns = [
    'income',
    'education_years',
    'credit_score',
    'debt',
    'study_hours',
    'attendance_pct'
]

imputer = SimpleImputer(strategy='mean')

dataset[numeric_columns] = imputer.fit_transform(
    dataset[numeric_columns]
)
imputer
```

Encoding Categorical Data with changing the gender like male 1 female 0

```python
from sklearn.preprocessing import LabelEncoder, OneHotEncoder

labelencoder_X = LabelEncoder()

X['gender'] = labelencoder_X.fit_transform(X['gender'])
labelencoder_X
```

```python
onehotencoder = OneHotEncoder(
    handle_unknown='ignore',
    sparse_output=False
)

X_city = onehotencoder.fit_transform(X[['city']])
X_city
```
